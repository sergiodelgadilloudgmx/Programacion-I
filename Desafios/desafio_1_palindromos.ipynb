Un palíndromo es una palabra o frase que se lee igual de adelante hacia atrás que de atrás hacia adelante, ignorando espacios, puntuación y mayúsculas.

**Idea:** se limpia el texto (solo letras y números, en minúsculas y sin acentos) y se compara con su reverso.

```python
import unicodedata


def limpiar(texto):
    """Deja solo letras y números en minúsculas, sin acentos."""
    # Separa las letras de sus acentos (á -> a + ´) y quita los acentos
    texto = unicodedata.normalize("NFD", texto)
    texto = "".join(c for c in texto if unicodedata.category(c) != "Mn")
    # Se queda solo con letras y números
    return "".join(c.lower() for c in texto if c.isalnum())


def es_palindromo(texto):
    limpio = limpiar(texto)
    return limpio == limpio[::-1]
```

### Pruebas con ejemplos

```python
ejemplos = [
    "Anita lava la tina",
    "Dábale arroz a la zorra el abad",
    "¿Acaso hubo búhos acá?",
    "Hola mundo",
]

for frase in ejemplos:
    resultado = "SÍ" if es_palindromo(frase) else "NO"
    print(f'"{frase}" -> {resultado} es palíndromo')
```

### Casos de prueba

Si alguna prueba falla, Python marca un `AssertionError`.

```python
assert es_palindromo("Anita lava la tina")
assert es_palindromo("Dábale arroz a la zorra el abad")
assert es_palindromo("Neuquén")
assert es_palindromo("a")
assert not es_palindromo("Hola mundo")
assert not es_palindromo("Python")
print("Todas las pruebas del desafío 1 pasaron.")
```

### Con tu propia frase

Ejecuta la celda y escribe una frase.

```python
texto = input("Ingresa una palabra o frase: ")

if not limpiar(texto):
    print("No ingresaste letras ni números.")
elif es_palindromo(texto):
    print(f'"{texto}" SÍ es un palíndromo.')
else:
    print(f'"{texto}" NO es un palíndromo.')
```

## Conclusión

Limpiar el texto antes de comparar permite ignorar espacios, signos, mayúsculas y acentos con una sola comparación contra el reverso. Si la entrada no tiene letras ni números, el programa lo avisa en lugar de dar un resultado falso.
