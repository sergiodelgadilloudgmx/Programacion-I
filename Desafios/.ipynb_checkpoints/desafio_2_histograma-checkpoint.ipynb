A partir de una lista de números se imprime una barra de asteriscos por cada uno; la barra mide lo que vale el número.

**Idea:** para cada número `n` se imprime `'*' * n`, con una etiqueta alineada a la izquierda.

```python
MAX_ANCHO = 50  # largo máximo de una barra


def histograma(numeros, max_ancho=MAX_ANCHO):
    """Imprime una barra de asteriscos por cada número.

    Si el número más grande pasa de max_ancho, las barras se escalan
    para que quepan en pantalla.
    """
    if not numeros:
        print("La lista está vacía.")
        return

    mayor = max(numeros)
    escala = max_ancho / mayor if mayor > max_ancho else 1
    ancho = len(str(mayor))  # para alinear las etiquetas

    for n in numeros:
        largo = round(n * escala)
        if n > 0:
            largo = max(1, largo)  # un número positivo siempre se ve
        print(f"{n:>{ancho}} | {'*' * largo}")

    if escala != 1:
        print(f"\nEscala: cada * representa aprox. {1 / escala:.1f}")
```

### Pruebas con ejemplos

```python
histograma([3, 7, 1, 5])
print()
histograma([12, 4, 0, 9])
print()
histograma([120, 30, 75, 1])   # números grandes: se escalan
print()
histograma([])
```

### Con tu propia lista

Ejecuta la celda y escribe números enteros positivos separados por espacios o comas.

```python
def pedir_numeros():
    """Pide los números separados por espacios o comas y los valida."""
    while True:
        entrada = input("Ingresa números enteros positivos separados por espacios o comas: ")
        partes = entrada.replace(",", " ").split()

        if not partes:
            print("No ingresaste nada, intenta de nuevo.")
            continue

        try:
            numeros = [int(p) for p in partes]
        except ValueError:
            print("Solo se permiten números enteros, intenta de nuevo.")
            continue

        if any(n < 0 for n in numeros):
            print("No se permiten números negativos, intenta de nuevo.")
            continue

        return numeros


numeros = pedir_numeros()
print()
histograma(numeros)
```

## Conclusión

Multiplicar `'*'` por el valor de cada número resuelve las barras sin ciclos adicionales. Se validan los datos (vacíos, texto y negativos), las barras se escalan cuando hay números muy grandes y una lista vacía no genera error.
